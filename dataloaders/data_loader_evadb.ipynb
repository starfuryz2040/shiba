```python
create_db = """
CREATE DATABASE postgres_data
WITH ENGINE = 'postgres',
PARAMETERS = {
"user" : "postgres",
"password" : "postgres",
"host" : "localhost",
"port" : "5432",
"database" : "evadb_idnet"
}
;
"""
```

```python
drop_table = ("""
                USE postgres_data {
                    DROP TABLE IF EXISTS idnet
                    };
              """)
```

```python
create_table = """
    USE postgres_data {
create table IF NOT EXISTS idnet
(
    name             text,
    imagedata      text,
    prelabeled_fraud integer,
    id               text,
    person_name      text,
    address          text,
    birthday         text,
    gender           text,
    ethnicity        text,
    class            text,
    issue_date       text,
    expire_date      text,
    eye_color        text,
    hair_color       text,
    is_donor         boolean,
    is_veteran       boolean,
    license_number   text,
    height_inches    integer,
    weight_lbs       integer
)}
;

"""
```
