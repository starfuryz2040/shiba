```python
import evadb
```

```python
cursor = evadb.connect().cursor()
```

```python
create_db = """
CREATE DATABASE postgres_data
WITH ENGINE = 'postgres',
PARAMETERS = {
"user" : "postgres",
"password" : "postgres",
"host" : "localhost",
"port" : "5432",
"database" : "evadb_idnet"
}
;
"""
```

```python
create_table = """
    USE postgres_data {
create table IF NOT EXISTS toll_audit
(
    license_number             text,
    plate_number      text,
    photo_id        text,
    timestamp               text,
    toll_id      text,
    speed          integer,
    pass_device_type         text,
    driver_hair_color           text,
    driver_eye_color        text,
    driver_weight            text,
    toll_amount_usd       decimal(8,2)
)}
;

"""
```

```python
print(cursor.query(create_table).df())
#print(cursor.query(drop_table).df())
```

```python
load_query = """
USE postgres_data{
    COPY toll_audit
    FROM '/data_sources/join_output.csv'
    DELIMITER ','
    CSV HEADER
};
"""
```
