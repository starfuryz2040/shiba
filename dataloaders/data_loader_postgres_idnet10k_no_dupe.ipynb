```python
import urllib

import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.types import INTEGER, BIGINT, VARCHAR, TEXT

```

```python

password = urllib.parse.quote_plus("postgres")
db_connection_str = f'postgresql+psycopg2://postgres:{password}@localhost:5434/postgres'
# 1. Initialize the PostgreSQL connection engine
# Format: postgresql://username:password@host:port/database_name
engine = create_engine(db_connection_str)


```

```python
# 2. Read the Parquet file into memory
df = pd.read_parquet("../silver/idnet10k_no_dupe.parquet", engine="pyarrow")

```

```python
db_dtypes = {
    "address": TEXT(),
    "birthday": TEXT(),
    "class": TEXT(),
    "ethnicity": TEXT(),
    "expire_date": TEXT(),
    "eye_color": TEXT(),
    "gender": TEXT(),
    "hair_color": TEXT(),
    "height_inches": INTEGER(),
    "id": TEXT(),
    "imageData": TEXT(),
    "is_donor": TEXT(),
    "is_veteran": TEXT(),
    "issue_date":  TEXT(),
    "license_number":  TEXT(),
    "name": TEXT(),
    "person_name": TEXT(),
    "prelabeled_fraud": INTEGER(),
    "weight_lbs": INTEGER(),
}
```

```python
# 3. Append or create the table inside PostgreSQL
df.to_sql(
    name="idnet10k_no_dupe",
    con=engine,
    if_exists="replace",  # Options: 'fail', 'replace', 'append'
    index=False,          # Discard the DataFrame's numeric index row
    dtype=db_dtypes,
)
```
