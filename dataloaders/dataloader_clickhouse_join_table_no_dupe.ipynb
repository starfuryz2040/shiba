```python
import clickhouse_connect

from clickhouse_connect.driver.tools import insert_file
import pyarrow
import pandas as pd
```

```python
client = clickhouse_connect.get_client(host='localhost', username='default', password='')
```

```python
df = pd.read_parquet('../silver/alt_docs_x5_no_dupe.parquet')
```

```python
client.command("""
create table toll_audit_no_dupe
(
    license_number             text,
    plate_number      text,
    photo_id            text,
    timestamp               text,
    toll_id      text,
    speed          integer,
    pass_device_type         text,
    driver_hair_color           text,
    driver_eye_color        text,
    driver_weight            text,
    toll_amount_usd       decimal(8,2)

)
    ENGINE = MergeTree
ORDER BY license_number
SETTINGS allow_nullable_key = 1;
""")
```
