```python
import csv
import sys

```

```python
def sanitize_csv(input_path, output_path, bad_lines_to_skip):
    """
    Reads a large CSV line by line and copies all lines to a new file,
    skipping specific offending 1-based line numbers.
    """
    # Bump Python's internal limit to handle massive fields without crashing
    csv.field_size_limit(sys.maxsize)

    print(f"Opening {input_path}...")

    # Using errors='ignore' ensures the script won't crash if it hits broken binary data
    with open(input_path, 'r', encoding='utf-8', errors='ignore') as infile, \
         open(output_path, 'w', encoding='utf-8', newline='') as outfile:

        for current_line_num, line in enumerate(infile, start=1):
            # Skip the row if it's the broken row
            if current_line_num in bad_lines_to_skip:
                print(f"Skipped offending line number: {current_line_num}")
                continue

            # Write the clean row immediately to disk
            outfile.write(line)

    print(f"Done! Cleaned file saved to: {output_path}")

# --- Configuration ---
#INPUT_FILE = "idnet100k_no_dupe.csv"
#OUTPUT_FILE = "idnet100k_clean.csv"
#OFFENDING_LINES = {8766}  # Python set lookup is extremely fast O(1)
```
